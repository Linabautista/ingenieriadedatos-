```python
df['Workout Type'].value_counts().plot(kind='bar')

plt.title('Cantidad de entrenamientos por tipo')
plt.xlabel('Tipo de entrenamiento')
plt.ylabel('Cantidad')
plt.xticks(rotation=45)
plt.show()
```

```python
plt.hist(df['Calories Burned'].dropna())

plt.title('Distribución de calorías quemadas')
plt.xlabel('Calorías quemadas')
plt.ylabel('Frecuencia')

plt.show()
```

```python
plt.scatter(
    df['Workout Duration (Minutes)'],
    df['Calories Burned']
)

plt.title('Duración del entrenamiento vs calorías quemadas')
plt.xlabel('Duración del entrenamiento (minutos)')
plt.ylabel('Calorías quemadas')

plt.show()
```
